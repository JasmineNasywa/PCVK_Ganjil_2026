```python
canvas_int16 = np.zeros(shape=(500,500,3),dtype=np.int16)
canvas_int32 = np.zeros(shape=(500,500,3),dtype=np.int32)

print("Kotak int16")
print("dtype :", canvas_int16.dtype)
print("min   :", canvas_int16.min())
print("max   :", canvas_int16.max())
print("nbytes:", canvas_int16.nbytes)

print("Kotak int32")
print("dtype :", canvas_int32.dtype)
print("min   :", canvas_int32.min())
print("max   :", canvas_int32.max())
print("nbytes:", canvas_int32.nbytes)

plt.subplot(1,2,1) #supaya horizontal 1 baris, 2 kolom, posisi)
plt.imshow(canvas_int16)
plt.title('int16')
plt.subplot(1,2,2)
plt.imshow(canvas_int32)
plt.title('int32')

plt.show()
```

alasan kenapa memorinya berbeda adalah, int membutuhkan 2 byte untuk setiap elemen sedangkan int32 membutuhkan 4 byte tiap elemen

# Tugas Praktikum - Analisa

Pertanyaan yang diajukan :

Apakah pencerminan horizontal, vertikal, dan keduanya mengubah dimensi serta jumlah piksel pada sebuah citra digital?

```python
img = cv.imread("cicak.jpg")
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

horizontal1 = cv.flip(img_rgb, 1)
vertical1 = cv.flip(img_rgb, 0)
verhon1 = cv.flip(img_rgb, -1)

plt.subplot(1, 3, 1)
plt.imshow(horizontal1)
plt.title("Horizontal")

plt.subplot(1, 3, 2)
plt.imshow(vertical1)
plt.title("Vertikal")

plt.subplot(1, 3, 3)
plt.imshow(verhon1)
plt.title("Horizontal + Vertikal")

plt.show()

print("Ukuran gambar")
print("asli      :", img_rgb.shape)
print("horizontal:", horizontal1.shape)
print("vertikal  :", vertical1.shape)
print("keduanya  :", verhon1.shape)

print("Jumlah piksel")
print("asli      :", img_rgb.shape[0] * img_rgb.shape[1])
print("horizontal:", horizontal1.shape[0] * horizontal1.shape[1])
print("vertikal  :", vertical1.shape[0] * vertical1.shape[1])
print("keduanya  :", verhon1.shape[0] * verhon1.shape[1])
```

```python


def beri_label(nama_file, nama):
  img = cv.imread(nama_file)
  tinggi, lebar = img.shape[:2]

  tinggi_label = int(tinggi * 0.12)
  margin = int(lebar * 0.03)
  font_scale = tinggi * 0.002
  ketebalan = int(tinggi * 0.004)

  cv.rectangle(img,(0, tinggi - tinggi_label),(lebar, tinggi),(0, 0, 0),-1)

  posisi_teks = (margin,tinggi - int(tinggi_label * 0.35))

  cv.putText(img,nama,posisi_teks,cv.FONT_HERSHEY_SIMPLEX,font_scale,(255, 255, 255),ketebalan)
  return img


img1 = beri_label("kodok.jpg", "Kodok")
img2 = beri_label("kurakura.jpg", "Kura-kura")
img3 = beri_label("landak.jpg", "Landak")


plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv.cvtColor(img1, cv.COLOR_BGR2RGB))
plt.title("Kodok")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv.cvtColor(img2, cv.COLOR_BGR2RGB))
plt.title("Kura-kura")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv.cvtColor(img3, cv.COLOR_BGR2RGB))
plt.title("Landak")
plt.axis("off")

plt.show()
```
