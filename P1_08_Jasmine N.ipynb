# Langkah2
Baca Image dari URLs

Pada langkah ini kita akan membaca image dari URLs, menampilkannya menggunakan OpenCV, memperhatikan ada perbedaan ketika membuka image dalam RGB dam BGR, channel warna default OpenCV adalah BGR

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://i.pinimg.com/1200x/56/05/d2/5605d29811d0574bae28caf4895ecb06.jpg","https://i.pinimg.com/736x/5e/51/a9/5e51a9ddcb88f30feafa578443ddb1c0.jpg","https://i.pinimg.com/1200x/24/c1/a7/24c1a7c912a2589f401c758e0880bfd2.jpg"]
#Baca dan tampilkan image
#loop pada tiap url image, beberapa image disimpan pada list
for url in urls:
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image,image_2))
  cv2_imshow(final_frame)
  print('\n')
```

# Langkah 3
melihat ukuran file image

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi, " x ",lebar)
cv2_imshow(image_2)
```

# Langkah 4

mengakses pixel dengan memberi diagonal menyilang

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range (lebar):
 image_3[int((tinggi)/2),y] = [255, 255, 255]

final_frame = cv.hconcat((image_2,image_3))
cv2_imshow(final_frame)
```

# Tugas praktikum D2

```python
print ("Garis vertikal")
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for x in range (tinggi):
 image_3[x,int((lebar)/2)] = [255, 255, 255]

final_frame = cv.hconcat((image_2,image_3))
cv2_imshow(final_frame)
```

```python
print ("Garis diagonal")
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#\
for i in range(min(tinggi, lebar)):
    image_3[i, i] = [255, 255, 255]
#/
for i in range(min(tinggi, lebar)):
    image_3[i, lebar - 1 - i] = [255, 255, 255]

final_frame = cv.hconcat((image_2,image_3))
cv2_imshow(final_frame)
```

```python
print ("Garis dengan panjang tertentu")
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)


panjang_garis = 278


baris_tengah = int(tinggi / 2)
awal = int(lebar / 2) - int(panjang_garis / 2)
akhir = int(lebar / 2) + int(panjang_garis / 2)

start = (awal, baris_tengah)
end = (akhir, baris_tengah)


cv.line(image_3, start, end, (255, 255, 255), thickness=2)

final_frame = cv.hconcat((image_2,image_3))
cv2_imshow(final_frame)
```

```python
print ("Kotak terisi")
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

lebar_kotak = 200
tinggi_kotak = 150
x1 = int(lebar / 2) - int(lebar_kotak / 2)
y1 = int(tinggi / 2) - int(tinggi_kotak / 2)
x2 = int(lebar / 2) + int(lebar_kotak / 2)
y2 = int(tinggi / 2) + int(tinggi_kotak / 2)

startkotak = (x1, y1)
endkotak = (x2, y2)


cv.rectangle(image_3, startkotak, endkotak, (255, 255, 255), thickness=-1)

final_frame = cv.hconcat((image_2,image_3))
cv2_imshow(final_frame)
```
